```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

BASE = "/kaggle/input/competitions/umud-challenge-muscle-architecture-in-ultrasound-data"

# --------------------------------------------------
# Paths
# --------------------------------------------------

image_path = f"{BASE}/apo_imgs_v1/apo_images_new_model_v1/image_0000.tif"
mask_path = f"{BASE}/apo_masks_v1/apo_masks_new_model_v1/image_0000.tif"

# --------------------------------------------------
# Load
# --------------------------------------------------

image = Image.open(image_path).convert("RGB")
mask = Image.open(mask_path).convert("L")

image_arr = np.array(image)
mask_arr = np.array(mask)

print("Image shape:", image_arr.shape)
print("Mask shape:", mask_arr.shape)

# --------------------------------------------------
# Resize mask ONLY for alignment investigation
# --------------------------------------------------

mask_resized = mask.resize(
    image.size,
    Image.Resampling.NEAREST
)

mask_resized_arr = np.array(mask_resized)

# --------------------------------------------------
# Create overlay
# --------------------------------------------------

overlay = image_arr.copy()

# Make annotation pixels visible
annotation = mask_resized_arr > 127

# Put white annotation pixels on top
overlay[annotation] = [255, 0, 0]

# --------------------------------------------------
# Plot
# --------------------------------------------------

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(image_arr)
axes[0].set_title("Original Ultrasound")

axes[1].imshow(mask_arr, cmap="gray")
axes[1].set_title("Original Aponeurosis Mask")

axes[2].imshow(overlay)
axes[2].set_title("Diagnostic Overlay\n(resized mask)")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

BASE = "/kaggle/input/competitions/umud-challenge-muscle-architecture-in-ultrasound-data"

image_folder = f"{BASE}/fasc_imgs_v1/fasc_images_new_model_v1"
mask_folder = f"{BASE}/fasc_masks_v1/fasc_masks_new_model_v1"

# Find files that have both image and mask
common_files = sorted(
    set(os.listdir(image_folder)) &
    set(os.listdir(mask_folder))
)

# Fixed seed so we can reproduce the same samples
random.seed(42)
samples = random.sample(common_files, 6)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for ax, filename in zip(axes.flat, samples):

    image = Image.open(
        os.path.join(image_folder, filename)
    ).convert("RGB")

    mask = Image.open(
        os.path.join(mask_folder, filename)
    ).convert("L")

    # Resize only for diagnostic visualization
    mask_resized = mask.resize(
        image.size,
        Image.Resampling.NEAREST
    )

    image_arr = np.array(image)
    mask_arr = np.array(mask_resized)

    overlay = image_arr.copy()

    annotation = mask_arr > 127
    overlay[annotation] = [255, 0, 0]

    ax.imshow(overlay)
    ax.set_title(filename)
    ax.axis("off")

plt.suptitle(
    "Fascicle Annotation Alignment — 6 Random Samples",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
import os
from collections import Counter
from PIL import Image

BASE = "/kaggle/input/competitions/umud-challenge-muscle-architecture-in-ultrasound-data"

image_folder = f"{BASE}/fasc_imgs_v1/fasc_images_new_model_v1"
mask_folder = f"{BASE}/fasc_masks_v1/fasc_masks_new_model_v1"

common_files = sorted(
    set(os.listdir(image_folder)) &
    set(os.listdir(mask_folder))
)

size_pairs = Counter()

for filename in common_files:
    image = Image.open(os.path.join(image_folder, filename))
    mask = Image.open(os.path.join(mask_folder, filename))

    size_pairs[(image.size, mask.size)] += 1

print("Total paired samples:", len(common_files))
print("Unique image/mask size combinations:", len(size_pairs))
print("=" * 80)

for (image_size, mask_size), count in size_pairs.most_common():
    print(
        f"Image {image_size[0]}x{image_size[1]:<5} | "
        f"Mask {mask_size[0]}x{mask_size[1]:<5} | "
        f"Count: {count}"
    )
```

```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

BASE = "/kaggle/input/competitions/umud-challenge-muscle-architecture-in-ultrasound-data"

image_folder = f"{BASE}/fasc_imgs_v1/fasc_images_new_model_v1"
fasc_mask_folder = f"{BASE}/fasc_masks_v1/fasc_masks_new_model_v1"
apo_mask_folder = f"{BASE}/apo_masks_v1/apo_masks_new_model_v1"

# Files that have all three:
# image + fascicle mask + aponeurosis mask
image_files = set(os.listdir(image_folder))
fasc_files = set(os.listdir(fasc_mask_folder))
apo_files = set(os.listdir(apo_mask_folder))

common = sorted(image_files & fasc_files & apo_files)

print("Images with image + fascicle mask + aponeurosis mask:", len(common))

random.seed(42)
samples = random.sample(common, 6)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for ax, filename in zip(axes.flat, samples):

    # Load image
    image = Image.open(
        os.path.join(image_folder, filename)
    ).convert("RGB")

    # Load masks
    fasc_mask = Image.open(
        os.path.join(fasc_mask_folder, filename)
    ).convert("L")

    apo_mask = Image.open(
        os.path.join(apo_mask_folder, filename)
    ).convert("L")

    # Resize masks ONLY for diagnostic visualization
    fasc_mask = fasc_mask.resize(
        image.size,
        Image.Resampling.NEAREST
    )

    apo_mask = apo_mask.resize(
        image.size,
        Image.Resampling.NEAREST
    )

    image_arr = np.array(image)
    fasc_arr = np.array(fasc_mask)
    apo_arr = np.array(apo_mask)

    overlay = image_arr.copy()

    # Fascicles → red
    fasc_pixels = fasc_arr > 127
    overlay[fasc_pixels] = [255, 0, 0]

    # Aponeuroses → green
    apo_pixels = apo_arr > 127
    overlay[apo_pixels] = [0, 255, 0]

    ax.imshow(overlay)
    ax.set_title(filename)
    ax.axis("off")

plt.suptitle(
    "Fascicle + Aponeurosis Annotation Overlay",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

BASE = "/kaggle/input/competitions/umud-challenge-muscle-architecture-in-ultrasound-data"

image_path = f"{BASE}/fasc_imgs_v1/fasc_images_new_model_v1/image_0000.tif"
fasc_path = f"{BASE}/fasc_masks_v1/fasc_masks_new_model_v1/image_0000.tif"
apo_path = f"{BASE}/apo_masks_v1/apo_masks_new_model_v1/image_0000.tif"

# --------------------------------------------------
# Load
# --------------------------------------------------

image = Image.open(image_path).convert("RGB")

fasc = Image.open(fasc_path).convert("L")
apo = Image.open(apo_path).convert("L")

# Diagnostic resize to image coordinate space
fasc = fasc.resize(image.size, Image.Resampling.NEAREST)
apo = apo.resize(image.size, Image.Resampling.NEAREST)

fasc_arr = np.array(fasc)
apo_arr = np.array(apo)

# --------------------------------------------------
# Extract annotation coordinates
# --------------------------------------------------

fasc_y, fasc_x = np.where(fasc_arr > 127)
apo_y, apo_x = np.where(apo_arr > 127)

print("Image size:", image.size)

print("\nFascicle annotation pixels:", len(fasc_x))
print(
    "Fascicle bounding box:",
    f"x={fasc_x.min()}..{fasc_x.max()},",
    f"y={fasc_y.min()}..{fasc_y.max()}"
)

print("\nAponeurosis annotation pixels:", len(apo_x))
print(
    "Aponeurosis bounding box:",
    f"x={apo_x.min()}..{apo_x.max()},",
    f"y={apo_y.min()}..{apo_y.max()}"
)

# --------------------------------------------------
# Visualize extracted pixels
# --------------------------------------------------

plt.figure(figsize=(12, 7))

plt.scatter(
    fasc_x,
    fasc_y,
    s=1,
    label="Fascicle pixels"
)

plt.scatter(
    apo_x,
    apo_y,
    s=1,
    label="Aponeurosis pixels"
)

plt.gca().invert_yaxis()
plt.xlabel("X pixel")
plt.ylabel("Y pixel")
plt.title("Raw Annotation Coordinates — image_0000")
plt.legend()

plt.show()
```

```python
import numpy as np
from PIL import Image
from scipy import ndimage

BASE = "/kaggle/input/competitions/umud-challenge-muscle-architecture-in-ultrasound-data"

mask_path = f"{BASE}/fasc_masks_v1/fasc_masks_new_model_v1/image_0000.tif"

mask = np.array(
    Image.open(mask_path).convert("L")
)

binary = mask > 127

# --------------------------------------------------
# Find connected components
# --------------------------------------------------

labeled, num_components = ndimage.label(binary)

objects = ndimage.find_objects(labeled)

components = []

for component_id, slc in enumerate(objects, start=1):

    if slc is None:
        continue

    ys, xs = np.where(labeled[slc] == component_id)

    # Convert local coordinates to full-image coordinates
    y0 = slc[0].start
    x0 = slc[1].start

    xs = xs + x0
    ys = ys + y0

    # Ignore extremely tiny components
    if len(xs) < 10:
        continue

    # --------------------------------------------------
    # Fit y = m*x + b
    # --------------------------------------------------

    m, b = np.polyfit(xs, ys, 1)

    angle = np.degrees(np.arctan2(m, 1))

    components.append({
        "id": component_id,
        "pixels": len(xs),
        "x_min": xs.min(),
        "x_max": xs.max(),
        "y_min": ys.min(),
        "y_max": ys.max(),
        "slope": m,
        "angle_deg": angle
    })

# Sort by position
components.sort(key=lambda c: (c["y_min"], c["x_min"]))

print("Total connected components:", num_components)
print("Useful fascicle components:", len(components))
print("=" * 90)

for c in components:
    print(
        f"Component {c['id']:2d} | "
        f"pixels={c['pixels']:4d} | "
        f"x={c['x_min']:3d}..{c['x_max']:3d} | "
        f"y={c['y_min']:3d}..{c['y_max']:3d} | "
        f"angle={c['angle_deg']:7.2f}°"
    )
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from scipy import ndimage

BASE = "/kaggle/input/competitions/umud-challenge-muscle-architecture-in-ultrasound-data"

mask_path = f"{BASE}/fasc_masks_v1/fasc_masks_new_model_v1/image_0000.tif"

mask = np.array(
    Image.open(mask_path).convert("L")
)

binary = mask > 127

# Original connected components
_, original_count = ndimage.label(binary)

# Small morphological closing
structure = np.ones((3, 3), dtype=bool)
closed = ndimage.binary_closing(
    binary,
    structure=structure
)

# Connected components after closing
labeled, closed_count = ndimage.label(closed)

# Component sizes
component_sizes = np.bincount(labeled.ravel())[1:]
useful_sizes = component_sizes[component_sizes >= 10]

print("Original connected components:", original_count)
print("Components after 3x3 closing:", closed_count)
print("Components >= 10 pixels after closing:", len(useful_sizes))

print("\nComponent sizes >= 10 pixels:")
print(np.sort(useful_sizes)[::-1])

# --------------------------------------------------
# Visual comparison
# --------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

axes[0].imshow(binary, cmap="gray")
axes[0].set_title("Original Fascicle Mask")

axes[1].imshow(closed, cmap="gray")
axes[1].set_title("After 3×3 Morphological Closing")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from PIL import Image

BASE = "/kaggle/input/competitions/umud-challenge-muscle-architecture-in-ultrasound-data"

mask_path = f"{BASE}/fasc_masks_v1/fasc_masks_new_model_v1/image_0000.tif"

mask = np.array(
    Image.open(mask_path).convert("L")
)

binary = np.where(mask > 127, 255, 0).astype(np.uint8)

# Detect line segments
lines = cv2.HoughLinesP(
    binary,
    rho=1,
    theta=np.pi / 180,
    threshold=8,
    minLineLength=8,
    maxLineGap=3
)

print("Detected Hough lines:", 0 if lines is None else len(lines))

if lines is not None:

    detected = []

    for line in lines[:, 0]:
        x1, y1, x2, y2 = line

        dx = x2 - x1
        dy = y2 - y1

        length = np.sqrt(dx**2 + dy**2)
        angle = np.degrees(np.arctan2(dy, dx))

        detected.append(
            (length, angle, x1, y1, x2, y2)
        )

    # Longest first
    detected.sort(reverse=True)

    print("\nLongest detected lines:")
    print("=" * 80)

    for length, angle, x1, y1, x2, y2 in detected[:20]:

        print(
            f"length={length:6.2f} px | "
            f"angle={angle:7.2f}° | "
            f"({x1},{y1}) -> ({x2},{y2})"
        )

    # --------------------------------------------------
    # Visualization
    # --------------------------------------------------

    display = np.zeros(
        (*binary.shape, 3),
        dtype=np.uint8
    )

    display[binary > 0] = [255, 255, 255]

    for length, angle, x1, y1, x2, y2 in detected[:20]:

        cv2.line(
            display,
            (x1, y1),
            (x2, y2),
            (255, 0, 0),
            1
        )

    plt.figure(figsize=(12, 7))
    plt.imshow(display)
    plt.title("Detected Fascicle Line Segments")
    plt.axis("off")
    plt.show()
```
